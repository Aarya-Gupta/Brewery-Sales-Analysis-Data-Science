# Hypothesis 1

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Load dataset
data = pd.read_csv("/kaggle/input/brewery-operations-and-market-analysis-dataset/brewery_data_complete_extended.csv")

# Create a contingency table between Quality_Score and Beer_Style
contingency_table = pd.crosstab(data['Quality_Score'], data['Beer_Style'])

# Perform Chi-square test
chi2, p, dof, expected = chi2_contingency(contingency_table)
print("Chi-square test statistic:", chi2)
print("P-value:", p)
```

# Hypothesis 2

```python
from scipy.stats import f_oneway

# Group pH levels by Location
locations = data['Location'].unique()
location_groups = [data[data['Location'] == loc]['pH_Level'] for loc in locations]

# Perform ANOVA
f_stat, p_val = f_oneway(*location_groups)
print("ANOVA F-statistic:", f_stat)
print("P-value:", p_val)
```

# Hypothesis 3

```python
from scipy.stats import ttest_ind

# Calculate median fermentation time
median_fermentation_time = data['Fermentation_Time'].median()

# Split data into long and short fermentation groups
long_fermentation = data[data['Fermentation_Time'] > median_fermentation_time]['Total_Sales']
short_fermentation = data[data['Fermentation_Time'] <= median_fermentation_time]['Total_Sales']

# Perform two-sample t-test
t_stat, p_val = ttest_ind(long_fermentation, short_fermentation)
print("T-test statistic:", t_stat)
print("P-value:", p_val)
```

# Hypothesis 4

```python
from scipy.stats import pearsonr

# Load dataset and calculate correlation
alcohol_content = data['Alcohol_Content']
quality_score = data['Quality_Score']

# Perform Pearson correlation test
corr, p_val = pearsonr(alcohol_content, quality_score)
print("Correlation coefficient:", corr)
print("P-value:", p_val)

```

```python
from scipy.stats import ttest_ind

# Median split of Alcohol_Content
median_alcohol = data['Alcohol_Content'].median()
high_alcohol_sales = data[data['Alcohol_Content'] > median_alcohol]['Total_Sales']
low_alcohol_sales = data[data['Alcohol_Content'] <= median_alcohol]['Total_Sales']

# Perform two-sample t-test
t_stat, p_val = ttest_ind(high_alcohol_sales, low_alcohol_sales)
print("T-test statistic:", t_stat)
print("P-value:", p_val)
```

```python
import statsmodels.api as sm

# Set up linear regression model
X = data['Bitterness']
y = data['Quality_Score']
X = sm.add_constant(X)  # add constant for intercept

model = sm.OLS(y, X).fit()
print(model.summary())
```
